## ① 첫 줄

실행하면 셀 **바로 아래**에 결과가 붙습니다.

````python
print("안녕하세요")
````

## ② 상자에 담기

아래 셀은 실행해도 **아무것도 안 나옵니다.** 값을 넣기만 했으니까요 — **오류가 아닙니다.**

````python
name = "김하늘"
age = 17
````

담아 둔 값은 **다음 셀에서 그대로** 씁니다. 코랩은 셀끼리 기억을 나눠 씁니다.

````python
print(name, age)
````

## ③ 일부러 오류 내 보기

아래 셀을 실행하면 **빨간 글씨**가 나옵니다. 놀라지 마세요. 일부러 내는 것입니다.

**빨간 글씨는 맨 아랫줄만 읽으면 됩니다.**

````python
print(score)
````

`NameError: name 'score' is not defined`
→ **score 라는 이름을 모르겠다**는 뜻입니다.

셋 중 하나입니다. ① 철자가 틀렸다 ② 만드는 줄을 안 썼다 ③ **그 셀을 실행하지 않았다.**
아래처럼 먼저 만들어 주면 됩니다.

````python
score = 90
print(score)
````

## ④ 입력받기  ← 오늘의 고비

아래 셀을 실행하면 **입력칸이 생깁니다.** 나이를 넣고 **Enter** 를 누르세요.

````python
age = input("나이를 입력하세요: ")
print(age)
print(type(age))
````

숫자를 넣었는데 `<class 'str'>`, 즉 **글자**라고 나옵니다.
`input()` 이 돌려주는 값은 **언제나 글자**입니다. 그래서 아래는 오류가 납니다.

````python
print(age + 1)
````

`TypeError: can only concatenate str (not "int") to str`
→ **글자와 숫자는 못 더한다**는 뜻입니다.

`int()` 로 감싸 **숫자로 바꾸면** 해결됩니다. 이 한 줄이 2단원 실습의 출발점입니다.

````python
age = int(input("나이를 입력하세요: "))

print(age + 1)
print(f"내년에는 {age + 1}살이네요.")
````

## ⑤ 여러 값을 한 번에

2단원에서 다루는 **데이터**는 이 모양으로 시작합니다.
**첫 번째 칸이 `[0]`** 이라는 것만 기억하세요.

````python
scores = [88, 92, 45, 77, 61]

print(scores[0])
print(len(scores))
print(sum(scores) / len(scores))
````

마지막 줄은 **`print` 없이도** 결과가 나옵니다. 코랩의 습관입니다.

````python
sum(scores) / len(scores)
````
