## 협력적인 문제 해결 자세 기르기 (195쪽 그림 Ⅳ-3)
**모둠 구성 및 역할 분담 → 모둠별 주제 선정 → 수행 계획 수립 및 발표 → 과제 수행 및 문제 해결 → 결과 발표 및 결과물 전시 → 평가 및 피드백**

### 함께해 보기 — 역할 분담표 (195쪽)

````python
모둠 = {
    '모둠명': '바다지킴이',
    '비전': '해변 쓰레기를 AI로 분류해 수거를 돕는다',
    '규칙': ['맡은 일은 기한 안에', 'AI가 낸 답은 근거를 확인한 뒤 쓰기'],
    '역할': {'이끔이(모둠 관리 · 기획 · 발표)': '○○○', '챙김이(포트폴리오 · 발표 자료)': '○○○',
             '중심이(프로그래머 — 오렌지3 · 코랩)': '○○○', '모음이(데이터 수집)': '○○○'},
}
for k, v in 모둠.items():
    print(k, ':', v if not isinstance(v, dict) else '')
    if isinstance(v, dict):
        for r, n in v.items(): print('   ', r, '→', n)
print('※ 한 사람이 여러 역할을 맡아도 되지만, 각자 하나씩은 책임지는 것이 원칙(195쪽)')
````

````python
from sklearn import linear_model
regr = linear_model.LinearRegression()
model = regr.fit(X, y)                       # 과정 25 · 26
y_pred = model.predict(X)                    # 과정 27
print('X 에 대응하는 연도 :', y_pred.ravel().round(2))
print('GHI 10 이 되는 해 :', model.____([[10]]).ravel())   # 과정 28 교과서 2032.68
print('GHI  0 이 되는 해 :', model.predict([[0]]).ravel())  # 과정 29 교과서 2054.39
````

> GHI 10 이 되는 해가 **2032년** → 유엔의 2030년 기아 종식은 어렵다는 예측, GHI 0 은 **2054년**.

### 더 알아보기 — 기울기 · 절편 · 회귀식 그래프 (207쪽)

````python
print('기울기 :', model.coef_.ravel(), ' 절편 :', model.intercept_)
plt.scatter(X, y, label='data')
plt.plot(X, y_pred, 'r-', label='linear regression model')
plt.title('y = {:.4f} * x + {:.2f}'.format(regr.coef_[0][0], regr.intercept_[0]))
plt.xlabel('GHI'); plt.ylabel('Year'); plt.legend(); plt.show()
````

회귀식 **연도 = −2.171 × GHI + 2054.39** — GHI 가 1 줄어들 때마다 약 2.2년이 걸린다는 뜻입니다.
