## ① 데이터 불러오기 — 오렌지3 `File`
왼쪽 📁 폴더 아이콘을 눌러 우리 팀 파일(CSV · 엑셀 · 오렌지3 `.tab`)을 올린 뒤 파일 이름을 적으세요.
한글 파일(엑셀에서 저장한 cp949 포함)도 알아서 읽어요.

⚠ 코랩에 올린 파일은 **런타임이 끊기면 사라져요.** 다시 접속했으면 📁에 파일을 다시 올리고 **런타임 → 모두 실행**.

```python
# [오렌지3] File · 담당: ____
# ┌─ 👉 이 셀에서 내가 채울 곳 ────────────────────────
# │ ✎ 파일_이름 — 📁에 올린 파일 이름 그대로(따옴표 안)
# │ ✍ ① 아래쪽 print 줄 df.□□ — (행 수, 열 수)를 알려 주는 속성
# └────────────────────────────────────────
import pandas as pd
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)

파일_이름 = '____.csv'   # ✎ 📁에 올린 파일 이름 그대로 (예: '광주태양광발전_2025_시간별_날씨포함.csv')

# ── 아래는 그대로 두세요 ──
def 불러오기(이름):
    # 한글 CSV(utf-8 · cp949) · 엑셀 · 오렌지3 .tab 을 알아서 읽어요
    import os
    if not os.path.exists(이름):
        올린 = [f for f in os.listdir() if f.lower().endswith(('.csv', '.xlsx', '.xls', '.tab'))]
        raise FileNotFoundError(f'«{이름}» 파일이 없어요. 지금 📁에 있는 파일: {올린 or "없음(런타임이 끊겨 지워졌으면 다시 올리기)"} → 이름을 그대로 복사해 쓰세요.')
    if 이름.endswith(('.xlsx', '.xls')):
        return pd.read_excel(이름)
    if 이름.endswith('.tab'):
        머리 = pd.read_csv(이름, sep='\t', nrows=2, header=None, dtype=str).fillna('')
        표 = pd.read_csv(이름, sep='\t', skiprows=[1, 2])
        for 열, 종류 in zip(표.columns, 머리.iloc[0]):
            if 종류 in ('discrete', 'd'):
                표[열] = 표[열].astype(str)        # 오렌지3에서 «범주»였던 열은 글자로
        return 표
    for 인코딩 in ('utf-8-sig', 'cp949'):
        try:
            return pd.read_csv(이름, encoding=인코딩, thousands=',')   # «1,234» 같은 숫자도 숫자로
        except UnicodeDecodeError:
            pass
    raise ValueError('글자가 깨져 못 읽었어요 → 엑셀에서 «다른 이름으로 저장 › CSV UTF-8»로 다시 저장해 올리세요.')

df = 불러오기(파일_이름)
df.columns = df.columns.astype(str).str.strip()   # 열 이름 앞뒤 빈칸 지우기
print('불러온 행, 열:', df.____)   # ✍ 코드 빈칸 ① 표의 (행 수, 열 수)를 알려 주는 속성 — 힌트: s로 시작 · 괄호 없음
if len(df) > 50000:
    df = df.sample(50000, random_state=0)
    print('⚠ 행이 너무 많아 5만 행만 무작위로 골랐어요(코랩이 멈추지 않게). 오렌지3에서 Data Sampler로 줄였다면 그 수에 맞춰 바꾸세요.')
df.head()
```

```python
# [오렌지3] Data Table · 담당: ____  — 열 이름 · 빈칸 · 값 종류부터 확인
# ┌─ 👉 이 셀에서 내가 채울 곳 ────────────────────────
# │ ✍ ② print 줄 df.□□().sum() — 빈칸(결측치)이면 True로 바꾸는 함수
# └────────────────────────────────────────
# ✍ 코드 빈칸 ② 빈칸(결측치)이면 True로 바꾸는 함수 — 힌트: is로 시작하는 네 글자(교과서 Ⅱ-02)
print(pd.DataFrame({'자료형': df.dtypes.astype(str), '빈칸 수': df.____().sum(), '값 종류 수': df.nunique()}))
print()
print('열 이름(복사해서 ②에 쓰세요):', list(df.columns))
```

## ② 입력과 정답 고르기 · 전처리 — 오렌지3 `Select Columns` · `Impute` · `Continuize`
오렌지3에서 한 것과 **같게** 합니다. 열 이름은 위 목록에서 **복사**해 띄어쓰기까지 똑같이!
- `문제`: 정답이 글자 · 등급이면 `'분류'`, 숫자(양 · 개수 · 온도 …)를 맞히면 `'예측'`
- 정답을 **계산해서 만든 열**(예: 등급의 바탕 숫자 · 발전량 · PM25)은 입력에서 빼요. 정답을 미리 보는 셈이에요.

```python
# [오렌지3] Select Columns · 담당: ____
# ┌─ 👉 이 셀에서 내가 채울 곳 ────────────────────────
# │ ✎ 입력_열 · 정답_열 · 문제 — 위 «열 이름» 목록에서 복사
# │ ✍ ③ 맨 아래 print 줄 df[정답_열].□□() — 값마다 몇 개인지 세는 함수
# └────────────────────────────────────────
입력_열 = ['____', '____']   # ✎ features — 오렌지3에서 입력(특성)으로 둔 열 (몇 개든 쉼표로)
정답_열 = '____'             # ✎ target — 오렌지3에서 정답(레이블)으로 둔 열 하나
문제 = '____'                # ✎ '분류' 또는 '예측'

# ── 아래는 그대로 두세요: 흔한 실수를 막아 줘요 ──
없는_열 = [c for c in 입력_열 + [정답_열] if c not in df.columns]
if 없는_열:
    raise KeyError(f'데이터에 없는 열 이름: {없는_열} → 위 «열 이름» 목록에서 복사해 띄어쓰기까지 똑같이 쓰세요.')
if 정답_열 in 입력_열:
    입력_열 = [c for c in 입력_열 if c != 정답_열]
    print(f'⚠ 정답 열 «{정답_열}» 이 입력에도 있어서 뺐어요(정답을 보고 정답을 맞히는 셈이라).')
if 문제 not in ('분류', '예측'):
    raise ValueError("문제 는 '분류' 또는 '예측' 으로 써요.")
정답_숫자 = pd.api.types.is_numeric_dtype(df[정답_열])
if 문제 == '예측' and not 정답_숫자:
    raise ValueError(f'«{정답_열}» 은 글자 열이라 «예측»(숫자 맞히기)을 못 해요 → 문제 = \'분류\' 로 바꾸거나 숫자 정답 열을 고르세요.')
if 문제 == '분류' and 정답_숫자 and df[정답_열].nunique() > 20:
    print(f'⚠ «{정답_열}» 은 값 종류가 {df[정답_열].nunique()}개인 숫자 열이에요. 양 · 개수를 맞히는 거라면 문제 = \'예측\' 이 맞고, 등급을 맞히려면 등급 열을 정답으로 고르세요.')
df = df[입력_열 + [정답_열]]                  # 오렌지3 Select Columns 처럼 쓰는 열만 남겨요
print('입력', len(입력_열), '개:', 입력_열, '· 정답:', 정답_열, '·', 문제)
if 문제 == '분류':
    # ✍ 코드 빈칸 ③ 값마다 몇 개인지 세는 함수 — 힌트: value_로 시작
    print('정답 값과 개수:', df[정답_열].____().to_dict(), '← ④ «놓치면_안되는_값»은 이 중 하나')
```

```python
# [오렌지3] Impute / Preprocess · 담당: ____
# ┌─ 👉 이 셀에서 내가 채울 곳 ────────────────────────
# │ ✎ 결측_처리 · 중복_지우기 — 오렌지3에서 한 것과 같게(같으면 그대로)
# └────────────────────────────────────────
결측_처리 = '행 지우기'    # ✎ 오렌지3 Impute 와 같게: '행 지우기' 또는 '평균으로 채우기'
중복_지우기 = False        # ✎ 오렌지3에서 Unique(중복 행 지우기)를 썼다면 True

빈칸 = df.isna().sum()
print('열마다 빈칸 수:', 빈칸[빈칸 > 0].to_dict() or '없음')
처음 = len(df)
df = df.dropna(subset=[정답_열])                          # 정답이 빈 행은 배울 수 없어 늘 지워요
if 결측_처리 == '평균으로 채우기':
    df = df.fillna(df.mean(numeric_only=True))             # 숫자 열은 평균으로
    df = df.fillna(df.mode().iloc[0])                      # 글자 열은 가장 많은 값으로
df = df.dropna()                                           # 남은 빈칸이 있는 행 지우기 (쓰는 열만 봐요)
if 중복_지우기:
    df = df.drop_duplicates()
print(f'처리 전 {처음}행 → 처리 후 {len(df)}행')
if len(df) < 처음 * 0.7:
    print('⚠ 행이 30 % 넘게 사라졌어요. 빈칸이 많은 열을 입력에서 빼거나 결측_처리 = \'평균으로 채우기\' 를 써 보세요.')
```

```python
# [오렌지3] Continuize (글자 열 → 숫자) · 담당: ____
# ┌─ 👉 이 셀에서 내가 채울 곳 ────────────────────────
# │ ✎ 글자로_볼_열 — «번호 · 코드» 열이 있을 때만(없으면 [] 그대로)
# └────────────────────────────────────────
글자로_볼_열 = []   # ✎ 숫자지만 «번호 · 코드»인 열(예: '노선코드', '지역코드')이 있으면 여기에. 없으면 [] 그대로

# ── 아래는 그대로 두세요: 오렌지3가 자동으로 해 주던 일을 코드로 ──
def 글자로(v):        # 정답 값 1 · 1.0 · '1' 을 모두 '1' 로 맞춰요
    try:
        f = float(v)
        return str(int(f)) if f == int(f) else str(f)
    except (TypeError, ValueError):
        return str(v).strip()

def 숫자로(표, 알림=True):
    표 = 표.copy()
    for c in list(표.columns):
        if c in 글자로_볼_열:
            표[c] = 표[c].astype(str)
        elif 표[c].dtype == object or str(표[c].dtype) in ('string', 'category'):
            글 = 표[c].astype(str)
            if 글.str.match(r'^\d{4}[-./]\d{1,2}').mean() > 0.9:          # 날짜 · 시각 글자 열
                날짜 = pd.to_datetime(글, errors='coerce')
                표[c + '_월'], 표[c + '_요일'] = 날짜.dt.month, 날짜.dt.dayofweek
                if (날짜.dt.hour > 0).any():
                    표[c + '_시'] = 날짜.dt.hour
                표 = 표.drop(columns=c)
                if 알림: print(f'«{c}» 은 날짜라서 «{c}_월 · {c}_요일» 숫자 열로 바꿨어요.')
            elif 글.nunique() > 50 and 글.nunique() > len(글) * 0.05:      # 주소 · 이름 · 문장처럼 값이 다 다른 글자 열
                표 = 표.drop(columns=c)
                if 알림: print(f'⚠ «{c}» 은 값 종류가 {글.nunique()}개인 글자 열(주소 · 이름 같은 것)이라 뺐어요. 그대로 두면 0/1 열이 너무 많아져 배우지 못해요.')
    return pd.get_dummies(표)

X = 숫자로(df[입력_열])
y = df[정답_열]
if 문제 == '분류':
    y = y.map(글자로)            # 숫자 등급(0/1/2)도 글자 등급처럼
print('입력 모양(행, 열):', X.shape)
if X.shape[1] > 100:
    print(f'⚠ 입력이 {X.shape[1]}열로 늘었어요. 값 종류가 많은 글자 열이 있는지 위 Data Table 에서 확인하세요.')
```

## ③ 나누고 모델 정의하기 — 오렌지3 `Data Sampler` · 학습 위젯

```python
# [오렌지3] Test and Score (Test on test data) · 담당: ____
# ┌─ 👉 이 셀에서 내가 채울 곳 ────────────────────────
# │ ✎ 놓치면_안되는_값 — 분류면 ②에서 본 정답 값 하나(예측이면 None)
# │ ✍ ⑤ model.□□(X_train, y_train) — 학습시키는 함수
# │ ✍ ⑥ pred = model.□□(X_test) — 예측하는 함수
# └────────────────────────────────────────
놓치면_안되는_값 = None  # ✎ 분류라면 ②에서 본 정답 값 하나. 예) '미달' · '나쁨' · '피싱' · 숫자 등급이면 2 (예측이면 None 그대로)

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, mean_absolute_error, mean_squared_error, r2_score
if 문제 == '분류' and 놓치면_안되는_값 is not None:
    놓치면_안되는_값 = 글자로(놓치면_안되는_값)
    if 놓치면_안되는_값 not in set(y):
        raise ValueError(f'«{놓치면_안되는_값}» 은 정답 열에 없는 값이에요. 이 중 하나를 그대로 쓰세요: {sorted(set(y))}')
def 재현율(실제, 예측): return recall_score(실제, 예측, labels=[놓치면_안되는_값], average='macro', zero_division=0)

# ✍ 코드 빈칸 ⑤ · ⑥ — ⑤ 훈련 데이터로 «학습»하는 함수(f로 시작 · 세 글자) · ⑥ 테스트 데이터로 «예측»하는 함수(p로 시작) — 교과서 Ⅱ-05 모델 구현 절차
model.____(X_train, y_train)
pred = model.____(X_test)
import numpy as np
if 문제 == '분류':
    값들 = sorted(set(y_train) | set(y_test))
    표 = confusion_matrix(y_test, pred, labels=값들)
    print('🧮 혼동 행렬 (Confusion Matrix) — 행 = 실제 정답 · 열 = 모델의 예측 · 대각선(↘) = 맞힌 개수')
    print(pd.DataFrame(표, index=['실제 ' + v for v in 값들], columns=['예측 ' + v for v in 값들]))
    맞힘, 전체 = int(np.trace(표)), int(표.sum())
    점수 = 맞힘 / 전체
    print()
    print('📏 평가 지표 (분류)')
    print(f'정확도 (CA)        : {점수:.3f}   ← 대각선(맞힌 칸) {맞힘}개 ÷ 전체 {전체}개')
    if 놓치면_안되는_값 is None:
        print('→ 위 «놓치면_안되는_값» 을 채우면 재현율 · 정밀도 · F1 · AUC 도 나와요')
    else:
        v = 놓치면_안되는_값; k = 값들.index(v)
        잡음, 실제수, 예측수 = int(표[k, k]), int(표[k, :].sum()), int(표[:, k].sum())
        재, 정 = (잡음 / 실제수 if 실제수 else 0), (잡음 / 예측수 if 예측수 else 0)
        print(f'«{v}» 재현율 (Recall)  : {재:.3f}   ← «실제 {v}» 줄 {실제수}개 중 모델이 맞게 찾은 것 {잡음}개 (놓친 것 {실제수 - 잡음}개)')
        print(f'«{v}» 정밀도 (Precision): {정:.3f}   ← «예측 {v}» 칸 {예측수}개 중 진짜 «{v}» {잡음}개 (헛경보 {예측수 - 잡음}개)')
        print(f'«{v}» F1 점수 (F1)    : {(2 * 재 * 정 / (재 + 정) if 재 + 정 else 0):.3f}   ← 재현율과 정밀도를 함께 본 점수')
        try:
            from sklearn.metrics import roc_auc_score
            if hasattr(model, 'predict_proba'):
                확률 = model.predict_proba(X_test)[:, list(model.classes_).index(v)]
            else:
                확률 = model.decision_function(X_test)
                if np.ndim(확률) > 1: 확률 = 확률[:, list(model.classes_).index(v)]
                elif list(model.classes_).index(v) == 0: 확률 = -확률
            print(f'«{v}» AUC            : {roc_auc_score([t == v for t in y_test], 확률):.3f}   ← 0.5 = 찍기 수준 · 1 에 가까울수록 «{v}» 쪽을 나머지와 잘 가름')
        except Exception:
            print(f'«{v}» AUC            : (이 모델 · 데이터에서는 계산할 수 없어요 — 다른 지표로 봐요)')
else:
    오차 = np.asarray(y_test, dtype=float) - np.asarray(pred, dtype=float)
    mae, mse, 점수 = mean_absolute_error(y_test, pred), mean_squared_error(y_test, pred), r2_score(y_test, pred)
    print('📏 평가 지표 (예측 · 회귀) — 숫자가 작을수록 좋은 것: MAE · MSE · RMSE / 1에 가까울수록 좋은 것: R²')
    print(f'MAE  (평균 절대 오차)   : {mae:.3f}   ← 예측이 실제와 평균 {mae:.3f}만큼 어긋나요')
    print(f'MSE  (평균 제곱 오차)   : {mse:.3f}   ← 어긋남을 제곱해 평균 — 크게 틀린 것을 더 크게 봐요')
    print(f'RMSE (평균 제곱근 오차) : {mse ** 0.5:.3f}   ← MSE 의 제곱근 · 정답과 같은 단위')
    print(f'R²   (결정계수)         : {점수:.3f}   ← 1 = 완벽 · 0 = 평균으로 찍은 것과 같음 · 0 보다 작으면 평균보다 못함')
    print()
    print('🔎 테스트 데이터 앞 5행 — 실제 · 예측 · 오차(실제 − 예측)  ← 이 오차들의 평균 크기가 MAE')
    print(pd.DataFrame({'실제': np.asarray(y_test)[:5], '예측': np.round(np.asarray(pred, dtype=float)[:5], 2), '오차': np.round(오차[:5], 2)}).to_string(index=False))

# ── 점수가 거의 완벽하면: 정답을 알려 주는 열이 섞였는지 열 하나씩 점검 ──
if 점수 >= 0.99:
    의심 = []
    for c in X_train.columns:
        한열 = (DecisionTreeClassifier(max_depth=3) if 문제 == '분류' else DecisionTreeRegressor(max_depth=3)).fit(X_train[[c]], y_train)
        if 한열.score(X_test[[c]], y_test) >= 0.98:
            의심.append(c)
    if 의심 and len(의심) == X_train.shape[1] and len(의심) > 1:
        print(f'⚠ 점수가 {점수:.3f}로 거의 완벽하고, 입력 열 어느 것 하나만으로도 다 맞혀요. 실제보다 너무 깔끔한(실험 · 합성) 데이터일 수 있어요 → 보고서에 «실제 측정 데이터로 다시 확인해야 함»을 적어요.')
    elif 의심:
        print(f'⚠ 점수가 {점수:.3f}로 거의 완벽해요. 열 하나만으로도 98 % 넘게 맞히는 열: {의심} → 이 중 정답을 계산해 만든 열(예: 등급의 바탕 숫자)이 있으면 입력에서 빼고 다시 재요.')
    else:
        print(f'⚠ 점수가 {점수:.3f}로 거의 완벽해요. 원래 쉬운 문제인지, 같은 행이 훈련 · 테스트에 겹치지 않았는지 확인해요.')
```

## 📊 4. 기준선 · 비교표 — 요소 ③
> ✅ **11/12부터 기준선 비교는 하지 않아요** — 아래 Constant 칸은 «모두 실행» 때 저절로 돌아가니 그대로 두고, 비교표의 «기준선» 칸은 비워 둬도 돼요. 대신 3절 ④ 혼동 행렬 · 평가 지표를 7절 «평가»에 내 말로 써요.

**기준선**(오렌지3 `Constant`) = 데이터를 보지 않고 **늘 같은 답으로 찍었을 때**의 점수예요. 분류는 훈련 데이터에서 가장 많은 값으로, 예측은 훈련 데이터 평균으로 찍어요. 우리 모델이 «배운 게 있다»고 말하려면 이것보다 나아야 해요. 아래 칸을 실행하면 자동으로 계산돼요.

```python
# [오렌지3] Constant (기준선) · 담당: ____
if 문제 == '분류':
    base = DummyClassifier(strategy='most_frequent').fit(X_train, y_train)
    base_pred = base.predict(X_test)
    print(f'기준선 = 늘 «{base_pred[0]}» 이라고 찍기 (훈련 데이터에서 가장 많은 값)')
    기준, 우리 = accuracy_score(y_test, base_pred), accuracy_score(y_test, pred)
    print(f'정확도 (CA) — 기준선 {기준:.3f} · 우리 모델 {우리:.3f}')
    if 놓치면_안되는_값 is not None:
        print(f'«{놓치면_안되는_값}» 재현율 — 기준선 {재현율(y_test, base_pred):.3f} · 우리 모델 {재현율(y_test, pred):.3f}')
    더_나음 = 우리 > 기준 or (놓치면_안되는_값 is not None and 재현율(y_test, pred) > 재현율(y_test, base_pred))
else:
    base = DummyRegressor().fit(X_train, y_train)
    base_pred = base.predict(X_test)
    print(f'기준선 = 늘 훈련 데이터 평균 {base_pred[0]:.3f} 으로 찍기')
    기준, 우리 = mean_absolute_error(y_test, base_pred), mean_absolute_error(y_test, pred)
    print(f'MAE (평균 절대 오차) — 기준선 {기준:.3f} · 우리 모델 {우리:.3f}  (작을수록 좋아요)')
    print(f'RMSE (평균 제곱근 오차) — 기준선 {mean_squared_error(y_test, base_pred) ** 0.5:.3f} · 우리 모델 {mean_squared_error(y_test, pred) ** 0.5:.3f}')
    더_나음 = 우리 < 기준
if not 더_나음:
    print('⚠ 우리 모델이 기준선보다 낫지 않아요. 한쪽 값이 아주 많은 데이터라면 정확도 대신 «놓치면 안 되는 값» 재현율을 보고, 입력 열 · 모델을 바꿔 보는 것이 5절 개선 실험 거리예요.')
```
