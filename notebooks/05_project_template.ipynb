## ② 전처리 · 입력과 정답 고르기 — 오렌지3 `Impute` · `Select Columns`

```python
# [오렌지3] Impute / Preprocess · 담당: ____
print('처리 전 행 수:', len(df))
df = df.dropna()            # 결측치가 있는 행 지우기 (오렌지3에서 평균으로 채웠다면 df.fillna(df.mean(numeric_only=True)) )
df = df.drop_duplicates()   # 중복 행 지우기
print('처리 후 행 수:', len(df))
```

```python
# [오렌지3] Select Columns · 담당: ____
X = df[['____', '____']]    # 입력(특성) — 오렌지3에서 features 로 둔 열
y = df['____']              # 정답(레이블) — 오렌지3에서 target 으로 둔 열
print('정답 열에 있는 값:', sorted(y.astype(str).unique())[:10])   # 분류라면 ④에서 이 값 중 하나를 그대로 써요
```

## ④ 학습하고 점수 재기 — 오렌지3 `Test and Score` (Test on test data)
맨 위 두 줄(`문제`, `놓치면_안되는_값`)만 우리 팀에 맞게 바꾸고 실행해요. **런타임 → 모두 실행** 해도 오류가 나지 않게 한 칸으로 합쳤어요.

```python
# [오렌지3] Test and Score (Test on test data) · 담당: ____
문제 = '분류'          # 정답이 글자 · 등급이면 '분류', 숫자(양 · 개수 · 온도 …)면 '예측' 으로 바꾸기
놓치면_안되는_값 = None  # 분류라면 ②에서 본 정답 값 하나를 따옴표째로. 예) '미달' · '나쁨' · '피싱' (띄어쓰기까지 똑같이)

model.fit(X_train, y_train)
pred = model.predict(X_test)
if 문제 == '분류':
    from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix
    print('정확도 CA :', accuracy_score(y_test, pred))
    if 놓치면_안되는_값 is None:
        print('→ 위 «놓치면_안되는_값» 을 채우면 재현율 · 정밀도도 나와요')
    else:
        print('재현율    :', recall_score(y_test, pred, pos_label=놓치면_안되는_값))   # 실제 그 값 중 모델이 찾아낸 비율
        print('정밀도    :', precision_score(y_test, pred, pos_label=놓치면_안되는_값))
    # [오렌지3] Confusion Matrix — 행 = 실제, 열 = 예측
    print(confusion_matrix(y_test, pred))
else:
    from sklearn.metrics import mean_squared_error, r2_score
    print('RMSE :', mean_squared_error(y_test, pred) ** 0.5)
    print('R²   :', r2_score(y_test, pred))
```

## 📊 4. 기준선 · 비교표 — 요소 ③ · 📋 캔바 8쪽
**기준선** = 모두 가장 많은 값으로(예측이면 평균으로) 찍었을 때의 점수예요. 우리 모델은 이것보다 나아야 해요. 아래 칸을 실행하면 자동으로 계산돼요.

```python
# [오렌지3] Constant (기준선) · 담당: ____
if 문제 == '분류':
    from sklearn.dummy import DummyClassifier
    base = DummyClassifier(strategy='most_frequent').fit(X_train, y_train)
    base_pred = base.predict(X_test)
    print('기준선 정확도 :', accuracy_score(y_test, base_pred))
    if 놓치면_안되는_값 is not None:
        print('기준선 재현율 :', recall_score(y_test, base_pred, pos_label=놓치면_안되는_값, zero_division=0))
else:
    from sklearn.dummy import DummyRegressor
    base = DummyRegressor().fit(X_train, y_train)
    base_pred = base.predict(X_test)
    print('기준선 RMSE :', mean_squared_error(y_test, base_pred) ** 0.5)
```
